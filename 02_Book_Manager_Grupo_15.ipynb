# Ejercicio 03

> Puntos: 1

Crear un context manager para manejar las transacciones a la base de datos.

Archivo/s de trabajo:
- book_manager/database/connection.py


**Implementado en** `database/connection.py`: la clase `Transaccion` es un context manager (`__enter__` / `__exit__`) que abre una sesión, hace `commit` si el bloque termina correctamente, `rollback` si se lanza una excepción (que se vuelve a propagar) y siempre cierra la sesión. Se obtiene con `conexion.transaccion()` y es el que usan todos los repositorios.

````python
#@title Ejercicio 03 - Context manager de transacciones
from sqlalchemy import text

# Caso 1: el bloque termina bien -> commit automático
with conexion.transaccion() as sesion:
    resultado = sesion.execute(text("SELECT 'transacción confirmada' AS mensaje")).scalar()
print("Caso 1:", resultado)

# Caso 2: se lanza una excepción dentro del bloque -> rollback automático
try:
    with conexion.transaccion() as sesion:
        sesion.execute(text("SELECT 1"))
        raise RuntimeError("falla simulada dentro de la transacción")
except RuntimeError as error:
    print("Caso 2: rollback ejecutado ->", error)
````

# Ejercicio 04

> Puntos: 1

Crear las tablas del sistema con sus tipos de datos y relaciones.

Archivo/s de trabajo:
- book_manager/models/models.py


**Implementado en** `models/models.py`: una clase `*Model` por tabla heredando de `Base` (`DeclarativeBase`), con tipos de datos (`Integer`, `String`, `Numeric(14, 2)`, `Date`), claves primarias (simples y compuestas), claves foráneas (`ForeignKey`), restricciones de unicidad (`UniqueConstraint`) y relaciones (`relationship`) con cascada:

| Tabla | Columnas | Relaciones |
|---|---|---|
| `generos` | id, nombre (único), descripcion | 1 → N libros |
| `editoriales` | id, nombre (único), pais | 1 → N libros |
| `monedas` | id, codigo (único), nombre, simbolo | 1 → N precios, 1 → N tipos_cotizacion |
| `tipos_cotizacion` | id, nombre, descripcion, moneda_id (FK) | único (moneda_id, nombre); 1 → N cotizaciones |
| `libros` | id, isbn (único), titulo, autor, editorial_id (FK), genero_id (FK), anio | 1 → N precios, 1 → 1 stock (cascada) |
| `precios` | id, libro_id (FK), moneda_id (FK), valor | único (libro_id, moneda_id) |
| `stock` | libro_id (PK, FK), cantidad, cantidad_minima | |
| `cotizaciones` | tipo_id (PK, FK), fecha (PK), compra, venta | |

Normalización: `TipoCotizacion` incorpora `moneda_id` (también en la entidad), de modo que un tipo de cambio pertenece a una moneda (el dólar tiene Oficial, Blue, MEP...; el euro o el real tienen su tipo Oficial).

````python
#@title Ejercicio 04 - Creación de las tablas
from sqlalchemy import inspect

conexion.crear_tablas()
inspector = inspect(conexion.engine)
for tabla in conexion.nombres_tablas():
    columnas = ", ".join(f"{c['name']}:{c['type']}" for c in inspector.get_columns(tabla))
    claves = ", ".join(f"{fk['constrained_columns'][0]}->{fk['referred_table']}" for fk in inspector.get_foreign_keys(tabla))
    print(f"{tabla:<18} {columnas}")
    if claves:
        print(f"{'':<18} FK: {claves}")
````
