# Ejercicio 03

> Puntos: 3

Implementar un scraper (CuspideSpider) con scrapy que obtenga del sitio de Cúspide los siguientes datos:
- Precio.
- url de la img.
- Formas de pago y precio.
- Descripción detallada.

Solo vamos a trabajar con nuestros libros, es decir, solo debo buscar los 10 primeros resultados de cada búsqueda (por título). No puedo buscar libros que no tengo en mi catálogo.

Implementar:
 - Loaders
 - Pipelines


````python
#@title Instalación de Scrapy
!pip install -q scrapy openpyxl
````

**Implementado en** `scraper/` (proyecto Scrapy):

- `spiders/cuspide.py` – **`CuspideSpider`**: lee de la base los libros de **nuestro catálogo** (no se busca nada que no tengamos), busca cada uno por título en `https://cuspide.com/?s=<título>&post_type=product` y sigue los **10 primeros resultados**. De la ficha de cada producto extrae precio (en pesos y el valor en U$s que informa el sitio), URL de la imagen, formas de pago, descripción detallada (sinopsis), ISBN, título y autor. Si la búsqueda no devuelve nada se reintenta con el título simplificado (el buscador no encuentra *"Sapiens: De animales a dioses"* pero sí *"Sapiens De animales a dioses"*).
- `items.py` – `LibroCuspideItem`.
- `loaders.py` – **Item Loader** `LibroCuspideLoader`: procesadores de entrada que limpian textos, convierten precios con formato argentino (`$ 73.499,00` → `73499.0`), arman URLs absolutas y dejan solo el nombre de cada medio de pago; `TakeFirst` / `Join` / `Identity` como salida.
- `pipelines.py` – **Pipelines**: `ValidacionPipeline` (descarta items sin precio/título/URL), `FormasPagoPipeline` (une cada forma de pago con su precio), `CoincidenciaPipeline` (puntaje 0-100 por ISBN, título y autor) y `BaseDatosPipeline` (guarda en la tabla `precios_web` y marca el resultado elegido de cada libro).
- `runner.py` – el spider corre en un proceso aparte porque el reactor de Twisted no puede reiniciarse dentro del mismo kernel; `ServicioScraping.ejecutar_scraping` lo lanza y devuelve el resumen.

La diferencia de precios para las alertas (Ejercicio 04) se pide **antes** de ejecutar el scraper.

````python
#@title Ejercicio 03 - Ejecución del CuspideSpider
texto = input("Diferencia de precios (%) para generar alertas [10]: ").strip()
UMBRAL = float(texto.replace(",", ".")) if texto else 10.0

resultado, alertas = servicios.comparador.scraping_con_alertas(UMBRAL)
print(resultado)
if resultado.libros_sin_resultados:
    print("Sin resultados en Cúspide:",
          [servicios.libros.obtener(i).titulo for i in resultado.libros_sin_resultados])

print("\nResultado elegido para cada libro:")
for web in servicios.scraping.seleccionados():
    libro = servicios.libros.obtener(web.libro_id)
    print(f"\n[{libro.titulo}] -> {web.titulo} (posición {web.posicion}, coincidencia {web.coincidencia})")
    print(f"  Precio      : $ {web.precio:,.2f}  |  U$s {web.precio_usd}")
    print(f"  Imagen      : {web.url_imagen}")
    print(f"  Formas pago : {web.formas_pago[:3]} ...")
    print(f"  Descripción : {web.descripcion[:150]}...")
````

# Ejercicio 04

> Puntos: 1

Comparar el precio interno con el precio web y generar alertas.

Cada vez que se ejecuta el scraper se debe generar un archivo de alertas del tipo CSV que se debe poder descargar para analizar los precios. La diferencia de precios debe ser ingresada por el usuario antes de cada ejecución del scraper.


**Implementado en** `services/services.py`: `ServicioComparadorPrecios.comparar` toma, para cada libro, el resultado elegido en la web y lo compara con el **precio interno en pesos** (los libros con precio de lista en USD se pasan a pesos con la última cotización del dólar Oficial). `ComparacionPrecio` calcula la diferencia en pesos y en % del precio interno.

`generar_alertas_csv(umbral)` genera `reportes/alertas/alertas_<fecha>.csv` con los libros cuya diferencia (en valor absoluto) alcanza el **umbral ingresado por el usuario antes de ejecutar el scraper** (celda anterior / opción 13 del menú). `scraping_con_alertas(umbral)` ejecuta el scraper y genera el CSV de esa misma ejecución, de modo que **cada ejecución deja su archivo de alertas**.

````python
#@title Ejercicio 04 - Comparación de precios y archivo de alertas
print(f"Umbral: {alertas.umbral_porcentual:.2f}%")
print(f"{'Libro':<32}{'Interno':>12}{'Web':>12}{'Diferencia':>12}{'Dif %':>9}  Alerta")
for c in alertas.comparaciones:
    marca = c.tipo_alerta if c.supera(UMBRAL) else ""
    print(f"{c.libro.titulo[:31]:<32}{c.precio_interno:>12,.2f}{c.precio_web.precio:>12,.2f}"
          f"{c.diferencia:>12,.2f}{c.diferencia_porcentual:>8.2f}%  {marca}")

print(f"\n{alertas}")
print("".join(open(alertas.archivo, encoding="utf-8-sig").readlines()[:6]))
files.download(str(alertas.archivo))  # descarga del CSV de alertas
````
